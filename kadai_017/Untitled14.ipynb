```python
# クラス定義
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def printinfo(self):
        print(f"名前：{self.name} 年齢：{self.age}")

    def check_adult(self):
        if self.age >= 20:
            print("大人")
        else:
            print("小人")


# 元データ（リスト）
user_name = ["花子", "太郎", "一郎"]
user_age  = [36, 18, 20]


# インスタンスを格納するリスト
humans = []

# リストから1人ずつインスタンス作成
for i in range(len(user_name)):
    human = Human(user_name[i], user_age[i])
    humans.append(human)

# 表示処理
for human in humans:
    human.printinfo()
    human.check_adult()
    print()
```
