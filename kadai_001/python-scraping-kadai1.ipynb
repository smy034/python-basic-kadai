```python
import requests
from bs4 import BeautifulSoup

# 対象URL
url = "https://news.yahoo.co.jp/expert/articles/8754dd10842abf628728b92a0b5d1f0ba3bad247"

# ヘッダー設定（重要：bot判定回避）
headers = {
    "User-Agent": "Mozilla/5.0"
}

# HTML取得
response = requests.get(url, headers=headers)

# ステータス確認
if response.status_code != 200:
    print("ページ取得に失敗しました")
    exit()

# BeautifulSoupで解析
soup = BeautifulSoup(response.text, "html.parser")

# 本文抽出（data-role="paragraph"）
paragraphs = soup.find_all("p", attrs={"data-role": "paragraph"})

# 本文を結合
article_text = ""
for p in paragraphs:
    article_text += p.get_text(strip=True) + "\n"

# 出力
print("■記事本文\n")
print(article_text)
```
